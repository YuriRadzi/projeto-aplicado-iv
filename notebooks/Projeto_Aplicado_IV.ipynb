# **Introdução**

A mobilidade urbana constitui um dos principais desafios enfrentados pelas grandes metrópoles, pois está diretamente relacionada ao acesso da população ao trabalho, à educação, à saúde e a outros serviços essenciais. Em uma cidade com as dimensões e a complexidade de São Paulo, compreender os padrões de deslocamento da população é importante para o desenvolvimento de sistemas de transporte mais eficientes, acessíveis e sustentáveis.

O transporte coletivo por ônibus possui papel relevante nesse contexto devido à sua ampla cobertura territorial e à quantidade de passageiros transportados diariamente. Entretanto, sua demanda varia ao longo do tempo em função de fatores como dias da semana, feriados, férias escolares, alterações tarifárias, condições econômicas e acontecimentos extraordinários.

Um exemplo importante dessas alterações ocorreu durante a pandemia de COVID-19, a partir de 2020, quando houve uma redução significativa dos deslocamentos urbanos. Posteriormente, a retomada gradual das atividades presenciais produziu novas mudanças nos padrões de utilização do transporte coletivo.

Intervenções públicas também podem modificar esse comportamento. Um exemplo recente é o programa Domingão Tarifa Zero, implantado no município de São Paulo em dezembro de 2023. Segundo a Prefeitura de São Paulo (2026), o programa atingiu a marca de 400 milhões de passageiros transportados gratuitamente em agosto de 2026. Na ocasião desse marco, foram registrados aproximadamente 2,83 milhões de passageiros, quantidade 17% superior à observada no primeiro domingo de agosto de 2023, antes da implantação do programa.

Nesse cenário, o projeto propõe analisar a evolução da demanda pelo transporte público por ônibus no município de São Paulo entre 2015 e 2026. A pesquisa utiliza informações públicas disponibilizadas pela São Paulo Transporte S.A. (SPTrans), referentes à quantidade diária de passageiros transportados pelo sistema municipal.

Os dados apresentam granularidade diária e incluem informações relacionadas ao total de passageiros transportados e às diferentes modalidades de pagamento. Como esses registros são disponibilizados separadamente pela instituição, foi desenvolvido um algoritmo para automatizar o processo de coleta, resultando em aproximadamente 3.650 arquivos. Esses arquivos serão consolidados em uma única base estruturada, passando por etapas de verificação, padronização, tratamento de dados ausentes e identificação de possíveis inconsistências.

A partir dessa base, o objetivo geral do projeto é desenvolver e avaliar modelos preditivos de séries temporais capazes de estimar a demanda futura de passageiros. Para atingir esse objetivo, serão realizadas etapas de coleta e consolidação dos dados, tratamento e preparação da base, análise exploratória, identificação de tendências, padrões sazonais e períodos atípicos, desenvolvimento de diferentes modelos de previsão e comparação de seu desempenho por meio de métricas adequadas.

A escolha do tema também se justifica pela possibilidade de aplicar técnicas de Ciência de Dados a um problema real e diretamente relacionado ao cotidiano da população paulistana. Além disso, a construção de uma série histórica consolidada permite observar padrões que dificilmente seriam identificados a partir da análise isolada dos arquivos disponibilizados pela SPTrans.

Do ponto de vista social, os resultados poderão contribuir para discussões relacionadas ao planejamento de linhas, dimensionamento da frota e adequação da oferta de transporte às necessidades da população. O projeto também apresenta relação com o Objetivo de Desenvolvimento Sustentável 11 (ODS 11), especialmente com a meta 11.2, que busca ampliar o acesso a sistemas de transporte seguros, acessíveis e sustentáveis.

```python
if os.path.isfile('passageiros_unificado.parquet'):
    caminho_arquivo = os.path.join('..', 'dados', 'passageiros_unificado.parquet')
else:
    caminho_arquivo = 'https://github.com/YuriRadzi/projeto-aplicado-iv/raw/refs/heads/main/dados/passageiros_unificado.parquet'

df_historico = pd.read_parquet(caminho_arquivo, engine='pyarrow')

df_historico.head()
```

```python
## dataset de feriados para enriquecimento de dados (https://github.com/joaopbini/feriados-brasil)
ano_inicial = 2015
ano_final = 2026

colunas_nacional = ['data', 'nome', 'tipo', 'descricao']
colunas_estadual = colunas_nacional + ['uf']
colunas_municipal = colunas_estadual + ['codigo_ibge']

estado = 'SP'
municipio = 3550308 # Código IBGE de São Paulo - SP

dados = []

for ano in range(ano_inicial, ano_final+1):
    arquivo_federal = f'https://github.com/joaopbini/feriados-brasil/raw/refs/heads/master/dados/feriados/nacional/csv/{ano}.csv'
    arquivo_estadual = f'https://github.com/joaopbini/feriados-brasil/raw/refs/heads/master/dados/feriados/estadual/csv/{ano}.csv'
    arquivo_municipal = f'https://github.com/joaopbini/feriados-brasil/raw/refs/heads/master/dados/feriados/municipal/csv/{ano}.csv'

    df_feriado_federal_ano = pd.read_csv(arquivo_federal, names=colunas_municipal,  header=None, )
    df_feriado_estadual_ano = pd.read_csv(arquivo_estadual, names=colunas_municipal, header=None, index_col=None)
    df_feriado_municipal_ano = pd.read_csv(arquivo_municipal, names=colunas_municipal, header=None)

    df_feriado_estadual_ano = df_feriado_estadual_ano[df_feriado_estadual_ano["uf"] == 'SP']

    df_feriado_municipal_ano = df_feriado_municipal_ano[df_feriado_municipal_ano['codigo_ibge'] == municipio]

    dados.append(df_feriado_federal_ano)
    dados.append(df_feriado_estadual_ano)
    dados.append(df_feriado_municipal_ano)


df_feriados = pd.concat(dados)

df_feriados["data"] = pd.to_datetime(df_feriados["data"], format='%d/%m/%Y').dt.floor('D')
df_feriados = df_feriados.sort_values(by=['data'])

# Padroniza a coluna de data no histórico e cria o indicador de feriado
feriados_idx = df_feriados['data'].drop_duplicates()
df_historico['data'] = pd.to_datetime(df_historico['data']).dt.floor('D')
df_historico['feriado'] = df_historico['data'].isin(feriados_idx)
```

```python
# Verificando valores ausentes
df_historico.isna().sum()
```

```python
# drop e correção de ausêntes
df_historico = df_historico.dropna(subset=['data', 'tipo', 'area', 'empresa', 'linha'])

colunas_int = df_historico.select_dtypes(include=['int64']).columns.tolist()

df_historico.loc[:, colunas_int] = df_historico[colunas_int].fillna(0)

# descarta linhas com poucos registros (Linhas muito novas ou descontinuadas no período)
count_linhas = df_historico["linha"].value_counts()
linhas_minimo_registros = count_linhas[count_linhas > 500].index

df_historico = df_historico[df_historico["linha"].isin(linhas_minimo_registros)]

# Adicionando o total de passageiros (Diferente de total de pagantes)
colunas_passageiro = colunas_int.copy()
colunas_passageiro.remove("total_pagantes")
colunas_passageiro.remove("ano_arquivo")
df_historico["total_passageiros"] = df_historico[colunas_passageiro].sum(axis=1)
```

```python
# Verificação de continuidade dos dados
inicio_analise = df_historico['data'].min()
fim_analise = df_historico['data'].max()

print(f"A análise inicia em {inicio_analise} e termina em {fim_analise}.")

dias_esperados = pd.date_range(start=inicio_analise, end=fim_analise, freq='D')
dias_existentes = pd.to_datetime(df_historico['data']).dt.floor('D').unique()
dias_ausentes = dias_esperados.difference(dias_existentes).to_list()

print(f"{len(dias_ausentes)} dias ausentes entre {inicio_analise} e {fim_analise}.")
print(dias_ausentes)

# Base original (não altera linhas existentes)
df_base = df_historico.copy()
df_base['data'] = pd.to_datetime(df_base['data']).dt.floor('D')
df_base['registro_inferido'] = False

# Base auxiliar para cálculo de estatísticas
df_calc = df_base.copy()
df_calc['dia_semana'] = df_calc['data'].dt.dayofweek
df_calc['mes'] = df_calc['data'].dt.month
feriados_idx = pd.to_datetime(df_feriados['data']).dt.floor('D').drop_duplicates()
df_calc['feriado'] = df_calc['data'].dt.floor('D').isin(feriados_idx)

chaves = ['tipo', 'area', 'empresa', 'linha']
colunas_num = df_base.select_dtypes(include=['Int64', 'int64', 'float64']).columns.tolist()
colunas_num = [c for c in colunas_num if c not in ['ano_arquivo']]

dias_ausentes_idx = pd.DatetimeIndex(dias_ausentes)
novos_registros = []

for ano in sorted(df_calc['data'].dt.year.unique()):
    datas_ausentes_ano = dias_ausentes_idx[dias_ausentes_idx.year == ano]
    if len(datas_ausentes_ano) == 0:
        continue

    df_ano = df_calc[df_calc['data'].dt.year == ano].copy()
    grupos = df_ano[chaves].drop_duplicates()

    datas_df = pd.DataFrame({'data': datas_ausentes_ano})
    datas_df['dia_semana'] = datas_df['data'].dt.dayofweek
    datas_df['mes'] = datas_df['data'].dt.month
    datas_df['feriado'] = datas_df['data'].dt.floor('D').isin(feriados_idx)

    # Cria somente linhas para datas realmente ausentes do ano
    base_novos = grupos.merge(datas_df, how='cross')

    for col in colunas_num:
        agg1 = (
            df_ano.groupby(chaves + ['mes', 'dia_semana', 'feriado'], observed=True)[col]
            .median()
            .round()
            .rename('m1')
            .reset_index()
        )
        agg2 = (
            df_ano.groupby(chaves + ['mes'], observed=True)[col]
            .median()
            .round()
            .rename('m2')
            .reset_index()
        )
        agg3 = (
            df_ano.groupby(chaves, observed=True)[col]
            .median()
            .round()
            .rename('m3')
            .reset_index()
        )

        serie_col = (
            base_novos[chaves + ['mes', 'dia_semana', 'feriado']]
            .merge(agg1, on=chaves + ['mes', 'dia_semana', 'feriado'], how='left')
            .merge(agg2, on=chaves + ['mes'], how='left')
            .merge(agg3, on=chaves, how='left')
        )

        base_novos[col] = (
            serie_col['m1']
            .fillna(serie_col['m2'])
            .fillna(serie_col['m3'])
            .fillna(0)
            .astype('Int64')
        )

    # Completa colunas não numéricas da nova linha
    base_novos['ano_arquivo'] = base_novos['data'].dt.year.astype('Int64')
    base_novos['registro_inferido'] = True
    if 'arquivo_origem' in df_base.columns:
        base_novos['arquivo_origem'] = 'imputacao_data_ausente'

    # Mantém apenas colunas originais da base
    for col in df_base.columns:
        if col not in base_novos.columns:
            base_novos[col] = pd.NA

    base_novos = base_novos[df_base.columns]
    novos_registros.append(base_novos)

if novos_registros:
    df_novos = pd.concat(novos_registros, ignore_index=True)
    df_novos = df_novos.drop_duplicates(subset=['data', 'linha'], keep='first')
else:
    df_novos = pd.DataFrame(columns=df_base.columns)

# Concatena sem alterar os registros originais
df_correcoes = pd.concat([df_base, df_novos], ignore_index=True)
df_correcoes = df_correcoes[df_correcoes['data'] <= fim_analise]
df_correcoes['data'] = pd.to_datetime(df_correcoes['data']).dt.floor('D')
df_correcoes = df_correcoes.sort_values(['data', 'tipo', 'area', 'empresa', 'linha'])


if 'registro_inferido' in df_correcoes.columns:
    df_correcoes['registro_inferido'] = df_correcoes['registro_inferido'].astype(bool)
```

```python
# Recalcula coluna total_passageiros para incluir dados inferidos
df_correcoes["total_passageiros"] = df_correcoes[colunas_passageiro].sum(axis=1)

# adiciona campos de data que serão usados com frequência
df_correcoes["dia_semana"] = df_correcoes['data'].dt.day_of_week
df_correcoes['ano_mes'] = df_correcoes['data'].dt.to_period('M').dt.to_timestamp()
```

```python
#Validação de continuidade dos dados para análise de distribuição dos meses

plot_df = df_correcoes.copy()


# totalização mensal das duas métricas
serie_mensal = (
    plot_df.groupby('ano_mes', as_index=False)[['total_passageiros', 'total_pagantes']]
    .sum()
)

serie_mensal_long = serie_mensal.melt(
    id_vars='ano_mes',
    value_vars=['total_passageiros', 'total_pagantes'],
    var_name='serie',
    value_name='valor'
)

plt.figure(figsize=(14, 7))
sns.lineplot(
    data=serie_mensal_long,
    x='ano_mes',
    y='valor',
    hue='serie',
    linewidth=2,
)

plt.title('Validação de continuidade nos valores de total mensal de passageiros e pagantes')
plt.xlabel('Data (mês/ano)')
plt.ylabel('Total mensal')
plt.xticks(rotation=45)
plt.legend(title='Série')
plt.tight_layout()
plt.show()
```

```python
df_tmp = df_correcoes.copy()
df_tmp = df_tmp.dropna(subset=["data", "linha"])
df_tmp["ano"] = df_tmp["data"].dt.year

resumo_anos = (
    df_tmp[df_tmp["ano"].isin([2016, 2018])]
    .groupby(["linha", "ano"], as_index=False)
    .agg(
        soma_passageiros=("total_passageiros", "sum"),
        total_registros=("linha", "size")
    )
)

# Pivot da soma de passageiros
piv_pass = (
    resumo_anos.pivot(index="linha", columns="ano", values="soma_passageiros")
       .rename(columns={2016: "passageiros_2016", 2018: "passageiros_2018"})
)

# Pivot da contagem de registros (nomes pedidos)
piv_regs = (
    resumo_anos.pivot(index="linha", columns="ano", values="total_registros")
       .rename(columns={2016: "total_2016", 2018: "total_2018"})
)


df_comparacoes_anos = (
    piv_pass.join(piv_regs, how="outer")
            .reset_index()
            .fillna(0)
)

for c in ["passageiros_2016", "passageiros_2018", "total_2016", "total_2018"]:
    df_comparacoes_anos[c] = df_comparacoes_anos[c].round().astype("Int64")


df_comparacoes_anos["perc_diferenca_passageiros"] = np.where(
    df_comparacoes_anos["passageiros_2016"] > 0,
    df_comparacoes_anos["passageiros_2018"] / df_comparacoes_anos["passageiros_2016"],
    np.nan
)

df_comparacoes_anos["perc_diferenca_registros"] = np.where(
    df_comparacoes_anos["total_2016"] > 0,
    df_comparacoes_anos["total_2018"] / df_comparacoes_anos["total_2016"],
    np.nan
)

df_comparacoes_anos = df_comparacoes_anos.sort_values(by='perc_diferenca_registros', ascending=False)

df_comparacoes_anos.head()
```

```python
df_chk = df_correcoes[["linha", "data"]].copy()
df_chk = df_chk.dropna(subset=["linha", "data"]).drop_duplicates(["linha", "data"])

# Janela global da análise
inicio_analise =  df_chk["data"].min()
fim_analise = df_chk["data"].max()
datas_esperadas = pd.date_range(inicio_analise, fim_analise, freq="D")
total_dias_periodo = len(datas_esperadas)

# Contagem de registros por ano (dias únicos por linha)
df_chk["ano"] = df_chk["data"].dt.year
reg_por_ano = (
    df_chk.groupby(["linha", "ano"])["data"]
    .nunique()
    .unstack(fill_value=0)
)

# Anos completos dentro da janela (evita punir ano parcial no início/fim)
anos_completos = [
    ano for ano in range(inicio_analise.year, fim_analise.year + 1)
    if pd.Timestamp(year=ano, month=1, day=1) >= inicio_analise
    and pd.Timestamp(year=ano, month=12, day=31) <= fim_analise
]

# Critério 1: pelo menos 365 registros em cada ano completo
if anos_completos:
    crit_365 = (reg_por_ano.reindex(columns=anos_completos, fill_value=0) >= 365).all(axis=1)
else:
    crit_365 = pd.Series(True, index=reg_por_ano.index)

# Critério 2: cobertura literal completa entre início e fim (sem faltar nenhum dia)
crit_cobertura_total = (
    df_chk.groupby("linha")["data"].nunique().eq(total_dias_periodo)
)

# Série final com as linhas que passam nos dois critérios
linhas_completas = pd.Series(
    sorted(crit_365.index[crit_365 & crit_cobertura_total]),
    name="linha"
)

linhas_completas
```

```python
#Engenharia de features para o treinamento/testes 

print(len(df_final))

df_final = df_final.sort_values(by=['linha', 'data']).reset_index(drop=True)

# ==========================================
# A. Atributos de Calendário
# ==========================================
df_final['mes'] = df_final['data'].dt.month
df_final['dia_do_mes'] = df_final['data'].dt.day
df_final['ano'] = df_final['data'].dt.year

# Férias escolares no Brasil (Janeiro e Julho)
df_final['ferias_escolares'] = df_final['mes'].isin([1, 7]).astype(int)


# ==========================================
# B. Defasagens (Lags)
# ==========================================
# Calculados por grupo de linha para não misturar históricos
group_linha = df_final.groupby('linha')['total_passageiros']

# Lags diários e semanais
df_final['lag_1d'] = group_linha.shift(1)   # Ontem
df_final['lag_7d'] = group_linha.shift(7)   # Mesmo dia da semana passada
df_final['lag_14d'] = group_linha.shift(14) # Mesmo dia há 2 semanas
df_final['lag_28d'] = group_linha.shift(28) # Mesmo dia há 4 semanas (aprox. 1 mês)


# ==========================================
# C. Estatísticas Móveis (Rolling Windows)
# ==========================================
# Importante: Usamos shift(1) antes do rolling para EVITAR vazamento do futuro (Data Leakage).
# A média móvel de hoje deve considerar apenas os dias anteriores a hoje.

df_final['media_movel_7d'] = (
    df_final.groupby('linha')['total_passageiros']
    .transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).mean())
)

df_final['media_movel_30d'] = (
    df_final.groupby('linha')['total_passageiros']
    .transform(lambda x: x.shift(1).rolling(window=30, min_periods=1).mean())
)

df_final['desvio_padrao_7d'] = (
    df_final.groupby('linha')['total_passageiros']
    .transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).std())
)


# ==========================================
# D. Tratar NaNs iniciais gerados pelos Shifts
# ==========================================
# Os primeiros 28 dias de cada linha terão valores nulos nos lags maiores.
# Caso use LightGBM/XGBoost, eles lidam bem com NaNs, mas você pode remover se preferir:
# df_final = df_final.dropna().reset_index(drop=True)
```

```python
colunas_treinamento_testes = ['data', 'tipo', 'area', 'empresa', 'linha',
       'feriado', 'total_passageiros',
       'dia_semana', 'mes', 'dia_do_mes',
       'ano', 'ferias_escolares', 'lag_1d', 'lag_7d', 'lag_14d', 'lag_28d',
       'media_movel_7d', 'media_movel_30d', 'desvio_padrao_7d']

df_treinamento = df_final[colunas_treinamento_testes].copy()

# Apagar os valores nulos das médias móveis (Primeiros 28 dias de cada linha)
df_treinamento = df_treinamento.dropna() 
```

```python
# Consolidação para NeuralProphet: garante 1 registro por linha e data
df_treinamento_np = df_treinamento.copy()
df_treinamento_np['data'] = pd.to_datetime(df_treinamento_np['data'])

dups_id_data = df_treinamento_np.duplicated(subset=['linha', 'data']).sum()
print(f'Duplicatas linha+data antes da consolidação: {dups_id_data}')

if dups_id_data > 0:
    agg_map = {
        'tipo': 'first',
        'area': 'first',
        'empresa': 'first',
        'feriado': 'max',
        'total_passageiros': 'sum',
        'dia_semana': 'first',
        'mes': 'first',
        'dia_do_mes': 'first',
        'ano': 'first',
        'ferias_escolares': 'first',
    }

    df_treinamento_np = (
        df_treinamento_np
        .groupby(['linha', 'data'], as_index=False)
        .agg(agg_map)
    )

# Recalcula lags e estatísticas móveis sobre a série consolidada
df_treinamento_np = df_treinamento_np.sort_values(['linha', 'data']).reset_index(drop=True)
grp = df_treinamento_np.groupby('linha')['total_passageiros']

df_treinamento_np['lag_1d'] = grp.shift(1)
df_treinamento_np['lag_7d'] = grp.shift(7)
df_treinamento_np['lag_14d'] = grp.shift(14)
df_treinamento_np['lag_28d'] = grp.shift(28)

df_treinamento_np['media_movel_7d'] = (
    grp.transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).mean())
)
df_treinamento_np['media_movel_30d'] = (
    grp.transform(lambda x: x.shift(1).rolling(window=30, min_periods=1).mean())
)
df_treinamento_np['desvio_padrao_7d'] = (
    grp.transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).std())
)

# Remove linhas iniciais sem histórico mínimo de lags
df_treinamento_np = df_treinamento_np.dropna().copy()

# NeuralProphet não aceita dtype pandas nullable (Int64), então convertemos para float64 padrão
colunas_numericas_np = [
    'feriado', 'total_passageiros', 'dia_semana', 'mes', 'dia_do_mes', 'ano',
    'ferias_escolares', 'lag_1d', 'lag_7d', 'lag_14d', 'lag_28d',
    'media_movel_7d', 'media_movel_30d', 'desvio_padrao_7d',
]
df_treinamento_np[colunas_numericas_np] = (
    df_treinamento_np[colunas_numericas_np]
    .apply(pd.to_numeric, errors='coerce')
    .astype('float64')
)

print(f'Tamanho após consolidação e recálculo: {len(df_treinamento_np)}')
print('Duplicatas linha+data depois:', df_treinamento_np.duplicated(subset=['linha', 'data']).sum())

# Sobrescreve a base usada na célula 32
df_treinamento = df_treinamento_np.copy()
```

```python
# Compatibilidade temporária para NeuralProphet + pandas
import pandas as pd
import numpy as np

# 1) NeuralProphet 0.9.0 usa Series.view() em alguns fluxos
if not hasattr(pd.Series, 'view'):
    def _series_view(self, dtype=None):
        if dtype is None:
            return self.to_numpy().view()
        return self.astype(dtype, copy=False)
    pd.Series.view = _series_view
    print('Patch aplicado: pd.Series.view')

# 2) Reset explícito de DataFrameGroupBy.apply para base do pandas
try:
    from pandas.core.groupby.generic import DataFrameGroupBy
    from pandas.core.groupby.groupby import GroupBy

    DataFrameGroupBy.apply = GroupBy.apply
    if hasattr(DataFrameGroupBy, '_np_orig_apply'):
        delattr(DataFrameGroupBy, '_np_orig_apply')
except Exception as exc:
    print(f'Não foi possível resetar DataFrameGroupBy.apply: {exc}')

# 3) Patch seguro para preservar coluna ID quando ela está no índice do resultado
try:
    from pandas.core.groupby.generic import DataFrameGroupBy

    DataFrameGroupBy._np_orig_apply = DataFrameGroupBy.apply

    def _apply_compat(self, func, *args, **kwargs):
        res = DataFrameGroupBy._np_orig_apply(self, func, *args, **kwargs)
        try:
            if isinstance(res, pd.DataFrame) and ('ID' not in res.columns):
                if isinstance(res.index, pd.MultiIndex) and ('ID' in res.index.names):
                    res = res.copy()
                    res['ID'] = res.index.get_level_values('ID')
        except Exception:
            pass
        return res

    DataFrameGroupBy.apply = _apply_compat
    print('Patch aplicado: compatibilidade de groupby.apply para coluna ID')
except Exception as exc:
    print(f'Não foi possível aplicar patch de groupby.apply: {exc}')
```

```python
# Base de modelagem com NeuralProphet + validação temporal e Optuna (opcional)
"""
Observação importante:
- Para evitar incompatibilidades entre pandas/neuralprophet no modo global multi-séries,
  este pipeline treina 1 modelo por linha (série individual).
- Assim, continuamos com previsões por linha e métricas globais + por linha.
"""

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

try:
    from neuralprophet import NeuralProphet
except ImportError as exc:
    raise ImportError(
        "NeuralProphet não encontrado. Instale com: pip install neuralprophet"
    ) from exc

try:
    import optuna
    OPTUNA_DISPONIVEL = True
except ImportError:
    OPTUNA_DISPONIVEL = False
    print("Optuna não está instalado. Busca de hiperparâmetros será ignorada.")

from sklearn.metrics import mean_absolute_error, mean_squared_error

# -----------------------------
# 1) Definições da base
# -----------------------------
df_modelo = df_treinamento.copy()
df_modelo = df_modelo.sort_values(['linha', 'data']).reset_index(drop=True)
df_modelo = df_modelo.dropna().copy()

regressoras = [
    'feriado', 'dia_semana', 'mes', 'dia_do_mes', 'ano', 'ferias_escolares',
    'lag_1d', 'lag_7d', 'lag_14d', 'lag_28d',
    'media_movel_7d', 'media_movel_30d', 'desvio_padrao_7d',
]

# Constrói uma série por linha no formato NeuralProphet
dados_por_linha = {}
for linha_id, g in df_modelo.groupby('linha', observed=True):
    serie = g[['data', 'total_passageiros'] + regressoras].copy()
    serie = serie.rename(columns={'data': 'ds', 'total_passageiros': 'y'})
    serie['ds'] = pd.to_datetime(serie['ds'])
    serie = serie.sort_values('ds').drop_duplicates(subset=['ds'], keep='first')
    serie = serie.dropna().copy()

    # Segurança: só mantém séries com histórico razoável
    if len(serie) >= 120:
        dados_por_linha[linha_id] = serie

print(f'Total de linhas elegíveis para modelagem: {len(dados_por_linha)}')

# Se quiser teste rápido, limite o número de linhas
LIMITE_LINHAS_TESTE = None  # ex.: 20 para teste rápido
if LIMITE_LINHAS_TESTE is not None:
    linhas_selecionadas = list(dados_por_linha.keys())[:LIMITE_LINHAS_TESTE]
    dados_por_linha = {k: dados_por_linha[k] for k in linhas_selecionadas}
    print(f'Modo teste rápido ativo. Linhas usadas: {len(dados_por_linha)}')

# -----------------------------
# 2) Cenários de validação
# -----------------------------
cutoff_train_a = pd.Timestamp('2024-12-31')
cutoff_test_a = pd.Timestamp('2025-06-30')

cutoff_train_b = pd.Timestamp('2025-12-31')
cutoff_test_b = pd.Timestamp('2026-06-30')

cenarios = [
    {'nome': 'A_2018_2024_test_2025_06', 'train_end': cutoff_train_a, 'test_end': cutoff_test_a},
    {'nome': 'B_2018_2025_test_2026_06', 'train_end': cutoff_train_b, 'test_end': cutoff_test_b},
]

# -----------------------------
# 3) Funções auxiliares
# -----------------------------
def criar_modelo_neuralprophet(params=None):
    base_params = dict(
        yearly_seasonality=True,
        weekly_seasonality=True,
        daily_seasonality=False,
        n_lags=28,
        n_forecasts=1,
        epochs=60,
        learning_rate=1.0,
        batch_size=128,
        loss_func='Huber',
    )
    if params:
        base_params.update(params)

    modelo = NeuralProphet(**base_params)
    for r in regressoras:
        modelo = modelo.add_lagged_regressor(r)
    return modelo


def preparar_split_serie(df_serie, train_end, test_end):
    treino = df_serie[df_serie['ds'] <= train_end].copy()
    teste = df_serie[(df_serie['ds'] > train_end) & (df_serie['ds'] <= test_end)].copy()

    if len(treino) < 90 or teste.empty:
        return None, None
    return treino, teste


def prever_no_teste(modelo, treino, teste):
    _ = modelo.fit(treino, freq='D', progress='none')
    fcst = modelo.predict(teste.copy())

    pred_col = 'yhat1' if 'yhat1' in fcst.columns else 'yhat'
    resultado = teste[['ds', 'y']].merge(
        fcst[['ds', pred_col]],
        on='ds',
        how='left'
    )
    resultado = resultado.rename(columns={pred_col: 'y_pred'})
    resultado = resultado.dropna(subset=['y_pred'])
    return resultado


def calcular_metricas(df_res):
    y_true = df_res['y']
    y_pred = df_res['y_pred']
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    denom = np.where(np.abs(y_true) < 1e-9, np.nan, np.abs(y_true))
    mape = np.nanmean(np.abs((y_true - y_pred) / denom)) * 100
    return {'MAE': mae, 'RMSE': rmse, 'MAPE_%': mape}


def avaliar_cenario_por_linha(dados_linhas, cenario, params_modelo=None):
    previsoes_lista = []
    metricas_linhas = []
    modelos_linhas = {}
    linhas_ignoradas = 0

    for linha_id, serie in dados_linhas.items():
        treino, teste = preparar_split_serie(serie, cenario['train_end'], cenario['test_end'])
        if treino is None:
            linhas_ignoradas += 1
            continue

        try:
            modelo = criar_modelo_neuralprophet(params=params_modelo)
            res = prever_no_teste(modelo, treino, teste)
            if res.empty:
                linhas_ignoradas += 1
                continue

            met = calcular_metricas(res)
            met['linha'] = linha_id
            met['n_treino'] = len(treino)
            met['n_teste'] = len(teste)
            metricas_linhas.append(met)

            res['linha'] = linha_id
            previsoes_lista.append(res)
            modelos_linhas[linha_id] = modelo
        except Exception:
            linhas_ignoradas += 1
            continue

    if not previsoes_lista:
        raise ValueError(f"Cenário {cenario['nome']}: nenhuma linha elegível para treino/teste.")

    previsoes_todas = pd.concat(previsoes_lista, ignore_index=True)
    metricas_globais = calcular_metricas(previsoes_todas)
    metricas_globais['cenario'] = cenario['nome']
    metricas_globais['linhas_treinadas'] = len(modelos_linhas)
    metricas_globais['linhas_ignoradas'] = linhas_ignoradas
    metricas_globais['n_amostras_teste_total'] = len(previsoes_todas)

    df_metricas_linha = pd.DataFrame(metricas_linhas).sort_values('RMSE')
    return metricas_globais, previsoes_todas, df_metricas_linha, modelos_linhas

# -----------------------------
# 4) Execução base (sem Optuna)
# -----------------------------
resultados_metricas = []
resultados_previsoes = {}
metricas_por_linha_cenario = {}
modelos_treinados = {}

for c in cenarios:
    met_global, prevs, met_linha, modelos = avaliar_cenario_por_linha(
        dados_por_linha, cenario=c, params_modelo=None
    )
    resultados_metricas.append(met_global)
    resultados_previsoes[c['nome']] = prevs
    metricas_por_linha_cenario[c['nome']] = met_linha
    modelos_treinados[c['nome']] = modelos

df_resultados_metricas = pd.DataFrame(resultados_metricas).sort_values('RMSE')
print('Resumo global dos cenários:')
display(df_resultados_metricas)

print('\nExemplo de métricas por linha (melhores RMSE do primeiro cenário):')
primeiro_cenario = cenarios[0]['nome']
display(metricas_por_linha_cenario[primeiro_cenario].head(10))

# -----------------------------
# 5) Busca de hiperparâmetros (opcional) com Optuna
# -----------------------------
RODAR_OPTUNA = False
N_TRIALS = 20

best_params_por_cenario = {}
optuna_resultados = {}

if RODAR_OPTUNA:
    if not OPTUNA_DISPONIVEL:
        raise ImportError('Optuna não está instalado. Rode: pip install optuna')

    for c in cenarios:
        print(f"\nIniciando Optuna para cenário: {c['nome']}")

        def objective(trial):
            params = {
                'epochs': trial.suggest_int('epochs', 40, 140, step=20),
                'learning_rate': trial.suggest_float('learning_rate', 0.01, 1.0, log=True),
                'batch_size': trial.suggest_categorical('batch_size', [64, 128, 256]),
                'n_lags': trial.suggest_categorical('n_lags', [14, 21, 28, 35]),
                'yearly_seasonality': trial.suggest_categorical('yearly_seasonality', [True, False]),
                'weekly_seasonality': trial.suggest_categorical('weekly_seasonality', [True, False]),
            }

            met_global, _, _, _ = avaliar_cenario_por_linha(dados_por_linha, cenario=c, params_modelo=params)
            return met_global['RMSE']

        study = optuna.create_study(direction='minimize')
        study.optimize(objective, n_trials=N_TRIALS, show_progress_bar=False)

        best_params_por_cenario[c['nome']] = study.best_params
        optuna_resultados[c['nome']] = study
        print(f"Melhor RMSE ({c['nome']}): {study.best_value:.4f}")
        print(f"Melhores parâmetros: {study.best_params}")

# -----------------------------
# 6) Treino final para previsão de +6 meses por linha
# -----------------------------
data_final_disponivel = df_modelo['data'].max()
horizonte_6m = data_final_disponivel + pd.DateOffset(months=6)
datas_futuras = pd.date_range(data_final_disponivel + pd.Timedelta(days=1), horizonte_6m, freq='D')

cenario_escolhido = cenarios[-1]['nome']
params_finais = best_params_por_cenario.get(cenario_escolhido, None)

previsoes_futuras = []
modelos_finais_por_linha = {}

for linha_id, serie in dados_por_linha.items():
    try:
        modelo_final = criar_modelo_neuralprophet(params=params_finais)
        _ = modelo_final.fit(serie, freq='D', progress='none')

        futuro = pd.DataFrame({'ds': datas_futuras})
        futuro['dia_semana'] = futuro['ds'].dt.dayofweek.astype(float)
        futuro['mes'] = futuro['ds'].dt.month.astype(float)
        futuro['dia_do_mes'] = futuro['ds'].dt.day.astype(float)
        futuro['ano'] = futuro['ds'].dt.year.astype(float)
        futuro['ferias_escolares'] = futuro['mes'].isin([1.0, 7.0]).astype(float)

        for col in ['feriado', 'lag_1d', 'lag_7d', 'lag_14d', 'lag_28d', 'media_movel_7d', 'media_movel_30d', 'desvio_padrao_7d']:
            futuro[col] = 0.0

        if 'df_feriados' in globals():
            feriados_set = set(pd.to_datetime(df_feriados['data']).dt.floor('D').tolist())
            futuro['feriado'] = futuro['ds'].dt.floor('D').isin(feriados_set).astype(float)

        fcst = modelo_final.predict(futuro)
        col_prev = 'yhat1' if 'yhat1' in fcst.columns else 'yhat'
        out = fcst[['ds', col_prev]].rename(columns={col_prev: 'y_pred'})
        out['linha'] = linha_id

        previsoes_futuras.append(out)
        modelos_finais_por_linha[linha_id] = modelo_final
    except Exception:
        continue

if previsoes_futuras:
    df_previsao_6m = pd.concat(previsoes_futuras, ignore_index=True)
    print(f"Previsões geradas para {df_previsao_6m['linha'].nunique()} linhas.")
    print(f"Horizonte final: {df_previsao_6m['ds'].max().date()}")
    display(df_previsao_6m.head())
else:
    df_previsao_6m = pd.DataFrame(columns=['ds', 'linha', 'y_pred'])
    print('Não foi possível gerar previsões futuras para as linhas com os filtros atuais.')

print("\nObservação: para melhorar a previsão futura de 6 meses, vale implementar atualização recursiva dos lags/médias no futuro.")
```
