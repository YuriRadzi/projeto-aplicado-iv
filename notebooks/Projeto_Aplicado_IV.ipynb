# **Referencial Teórico**

A previsão da demanda por ônibus envolve a análise de séries temporais, nas quais os registros de passageiros são organizados cronologicamente para identificar tendências, sazonalidades e mudanças no comportamento da população. Esses fatores precisam ser considerados na escolha dos modelos, principalmente quando acontecimentos excepcionais alteram padrões que antes eram relativamente estáveis.

Daroncho e Martinez (2024) investigaram a demanda de ônibus em São Paulo entre 2005 e 2023 utilizando o modelo estatístico SARIMA, que considera a dependência entre observações anteriores e os padrões sazonais da série. Os autores construíram o modelo com dados anteriores à pandemia e compararam as previsões com a demanda registrada posteriormente. Para 2023, encontraram uma previsão média de 206 milhões de passageiros mensais, enquanto o valor observado foi de 173 milhões. O resultado mostra a dificuldade de prever mudanças expressivas utilizando apenas o comportamento histórico da demanda.

Silva e Daroncho (2024) adotaram uma abordagem diferente, baseada na análise quantitativa dos registros diários da SPTrans entre 2015 e 2023. Em vez de desenvolver um modelo preditivo, examinaram as variações da demanda antes, durante e depois da pandemia. Identificaram que a redução no número de passageiros já ocorria antes de 2020 e que a recuperação posterior não restabeleceu os níveis anteriores, passando de aproximadamente 10 milhões para 6,5 milhões de passageiros diários.

Cyril, Mulangi e George (2019) estudaram a previsão da demanda de ônibus na Índia utilizando dados de bilhetagem eletrônica e quatro variações do método Holt-Winters. Foram comparados modelos aditivos e multiplicativos, com e sem amortecimento da tendência. O modelo multiplicativo com tendência amortecida apresentou o menor erro percentual absoluto médio (MAPE), de 8,29%. Embora aplicado a um sistema diferente do paulistano, o estudo demonstra a importância de testar abordagens distintas para representar a sazonalidade.

Triebe et al. (2021) desenvolveram o NeuralProphet, modelo que combina componentes de séries temporais com técnicas de aprendizado de máquina, permitindo considerar tendências, sazonalidades, autorregressão e variáveis externas. Essa abordagem foi escolhida pelo nosso grupo para estimar a demanda diária de passageiros por linha de ônibus, utilizando dados da SPTrans. Após a análise exploratória e o tratamento dos registros, selecionamos séries com maior continuidade histórica e criamos variáveis relacionadas a feriados, férias escolares, dias da semana, defasagens temporais e médias móveis. A modelagem foi estruturada individualmente por linha, considerando sazonalidades semanais e anuais. Também foram definidos dois períodos de validação temporal, com treinamento e teste em intervalos distintos, além das métricas MAE, RMSE e MAPE para avaliar a precisão das previsões. Com isso, buscamos verificar se o modelo consegue representar as diferenças de demanda entre as linhas e produzir estimativas úteis para o planejamento do transporte coletivo.

## Enriquecimento de dados com informações de feriados
Na célula abaixo esta sendo obtido uma listagem de feriados no Brasil, considerando feriados nacionais, do estado de SP e específicos do município de SP. Com estes dados é esperado enriquecer o dataset, adicionando a informação se determinado dia é um feriado.

```python
## dataset de feriados para enriquecimento de dados (https://github.com/joaopbini/feriados-brasil)
ano_inicial = 2015
ano_final = 2026

colunas_nacional = ['data', 'nome', 'tipo', 'descricao']
colunas_estadual = colunas_nacional + ['uf']
colunas_municipal = colunas_estadual + ['codigo_ibge']

estado = 'SP'
municipio = 3550308 # Código IBGE de São Paulo - SP

dados = []

for ano in range(ano_inicial, ano_final+1):
    arquivo_federal = f'https://github.com/joaopbini/feriados-brasil/raw/refs/heads/master/dados/feriados/nacional/csv/{ano}.csv'
    arquivo_estadual = f'https://github.com/joaopbini/feriados-brasil/raw/refs/heads/master/dados/feriados/estadual/csv/{ano}.csv'
    arquivo_municipal = f'https://github.com/joaopbini/feriados-brasil/raw/refs/heads/master/dados/feriados/municipal/csv/{ano}.csv'

    df_feriado_federal_ano = pd.read_csv(arquivo_federal, names=colunas_municipal,  header=None, )
    df_feriado_estadual_ano = pd.read_csv(arquivo_estadual, names=colunas_municipal, header=None, index_col=None)
    df_feriado_municipal_ano = pd.read_csv(arquivo_municipal, names=colunas_municipal, header=None)

    df_feriado_estadual_ano = df_feriado_estadual_ano[df_feriado_estadual_ano["uf"] == 'SP']

    df_feriado_municipal_ano = df_feriado_municipal_ano[df_feriado_municipal_ano['codigo_ibge'] == municipio]

    dados.append(df_feriado_federal_ano)
    dados.append(df_feriado_estadual_ano)
    dados.append(df_feriado_municipal_ano)


df_feriados = pd.concat(dados)

df_feriados["data"] = pd.to_datetime(df_feriados["data"], format='%d/%m/%Y').dt.floor('D')
df_feriados = df_feriados.sort_values(by=['data'])

# Padroniza a coluna de data no histórico e cria o indicador de feriado
feriados_idx = df_feriados['data'].drop_duplicates()
df_historico['data'] = pd.to_datetime(df_historico['data']).dt.floor('D')
df_historico['feriado'] = df_historico['data'].isin(feriados_idx) #Campo booleano que indica se o dia é um feriado ou não.
```

## Garantia de continuidade de dados
No dataset, existem 192 dias ausentes. Para correção desta informação foi feito um algoritmo para fazer a inferência destes valores considerando a mediana dos valores de passageiros agrupados pela linha dentro do mesmo mês e dia da semana.

```python
# Verificação de continuidade dos dados
inicio_analise = df_historico['data'].min()
fim_analise = df_historico['data'].max()

print(f"A análise inicia em {inicio_analise} e termina em {fim_analise}.")

dias_esperados = pd.date_range(start=inicio_analise, end=fim_analise, freq='D')
dias_existentes = pd.to_datetime(df_historico['data']).dt.floor('D').unique()
dias_ausentes = dias_esperados.difference(dias_existentes).to_list()

print(f"{len(dias_ausentes)} dias ausentes entre {inicio_analise} e {fim_analise}.")
print(dias_ausentes)

# Base original (não altera linhas existentes)
df_base = df_historico.copy()
df_base['data'] = pd.to_datetime(df_base['data']).dt.floor('D')
df_base['registro_inferido'] = False

# Base auxiliar para cálculo de estatísticas
df_calc = df_base.copy()
df_calc['dia_semana'] = df_calc['data'].dt.dayofweek
df_calc['mes'] = df_calc['data'].dt.month
feriados_idx = pd.to_datetime(df_feriados['data']).dt.floor('D').drop_duplicates()
df_calc['feriado'] = df_calc['data'].dt.floor('D').isin(feriados_idx)

chaves = ['tipo', 'area', 'empresa', 'linha']
colunas_num = df_base.select_dtypes(include=['Int64', 'int64', 'float64']).columns.tolist()
colunas_num = [c for c in colunas_num if c not in ['ano_arquivo']]

dias_ausentes_idx = pd.DatetimeIndex(dias_ausentes)
novos_registros = []

for ano in sorted(df_calc['data'].dt.year.unique()):
    datas_ausentes_ano = dias_ausentes_idx[dias_ausentes_idx.year == ano]
    if len(datas_ausentes_ano) == 0:
        continue

    df_ano = df_calc[df_calc['data'].dt.year == ano].copy()
    grupos = df_ano[chaves].drop_duplicates()

    datas_df = pd.DataFrame({'data': datas_ausentes_ano})
    datas_df['dia_semana'] = datas_df['data'].dt.dayofweek
    datas_df['mes'] = datas_df['data'].dt.month
    datas_df['feriado'] = datas_df['data'].dt.floor('D').isin(feriados_idx)

    # Cria somente linhas para datas realmente ausentes do ano
    base_novos = grupos.merge(datas_df, how='cross')

    for col in colunas_num:
        agg1 = (
            df_ano.groupby(chaves + ['mes', 'dia_semana', 'feriado'], observed=True)[col]
            .median()
            .round()
            .rename('m1')
            .reset_index()
        )
        agg2 = (
            df_ano.groupby(chaves + ['mes'], observed=True)[col]
            .median()
            .round()
            .rename('m2')
            .reset_index()
        )
        agg3 = (
            df_ano.groupby(chaves, observed=True)[col]
            .median()
            .round()
            .rename('m3')
            .reset_index()
        )

        serie_col = (
            base_novos[chaves + ['mes', 'dia_semana', 'feriado']]
            .merge(agg1, on=chaves + ['mes', 'dia_semana', 'feriado'], how='left')
            .merge(agg2, on=chaves + ['mes'], how='left')
            .merge(agg3, on=chaves, how='left')
        )

        base_novos[col] = (
            serie_col['m1']
            .fillna(serie_col['m2'])
            .fillna(serie_col['m3'])
            .fillna(0)
            .astype('Int64')
        )

    # Completa colunas não numéricas da nova linha
    base_novos['ano_arquivo'] = base_novos['data'].dt.year.astype('Int64')
    base_novos['registro_inferido'] = True
    if 'arquivo_origem' in df_base.columns:
        base_novos['arquivo_origem'] = 'imputacao_data_ausente'

    # Mantém apenas colunas originais da base
    for col in df_base.columns:
        if col not in base_novos.columns:
            base_novos[col] = pd.NA

    base_novos = base_novos[df_base.columns]
    novos_registros.append(base_novos)

if novos_registros:
    df_novos = pd.concat(novos_registros, ignore_index=True)
    df_novos = df_novos.drop_duplicates(subset=['data', 'linha'], keep='first')
else:
    df_novos = pd.DataFrame(columns=df_base.columns)


# Garantir colunas iguais e sem duplicatas
df_base = df_base.loc[:, ~df_base.columns.duplicated()].copy()
df_novos = df_novos.loc[:, ~df_novos.columns.duplicated()].copy()
df_novos = df_novos.reindex(columns=df_base.columns)

# Normalizar tipos críticos antes do concat
df_base['data'] = pd.to_datetime(df_base['data'], errors='coerce').astype('datetime64[ns]')
df_novos['data'] = pd.to_datetime(df_novos['data'], errors='coerce').astype('datetime64[ns]')

if 'registro_inferido' in df_base.columns:
    df_base['registro_inferido'] = df_base['registro_inferido'].astype('boolean')
if 'registro_inferido' in df_novos.columns:
    df_novos['registro_inferido'] = df_novos['registro_inferido'].astype('boolean')

# Concatena sem alterar os registros originais
df_correcoes = pd.concat([df_base, df_novos], ignore_index=True)
df_correcoes = df_correcoes[df_correcoes['data'] <= fim_analise]
df_correcoes['data'] = pd.to_datetime(df_correcoes['data']).dt.floor('D')
df_correcoes = df_correcoes.sort_values(['data', 'tipo', 'area', 'empresa', 'linha'])


if 'registro_inferido' in df_correcoes.columns:
    df_correcoes['registro_inferido'] = df_correcoes['registro_inferido'].astype(bool)
```

```python
# Recalcula coluna total_passageiros para incluir dados inferidos
df_correcoes["total_passageiros"] = df_correcoes[colunas_passageiro].sum(axis=1)

# adiciona campos de data que serão usados com frequência
df_correcoes["dia_semana"] = df_correcoes['data'].dt.day_of_week
df_correcoes['ano_mes'] = df_correcoes['data'].dt.to_period('M').dt.to_timestamp()
```

## Picos de desproporcionais de dados
Conforme visualizado no gráfico anterior, existe uma grande quantidade de picos de utilização fora do padrão esperado, na seguinte análise foi descoberto uma grande quantidade de registros duplicados e que antes de 2018 existe uma inconstância de registros, algumas linhas eram registradas apenas em dias esporádicos, não sendo possíveis serem utilizados, logo serão descartados todos os registros anteriores a 2018 por falta de qualidade de dados nestes período.

```python
df_tmp = df_correcoes.copy()
df_tmp = df_tmp.dropna(subset=["data", "linha"])
df_tmp["ano"] = df_tmp["data"].dt.year

resumo_anos = (
    df_tmp[df_tmp["ano"].isin([2016, 2018])]
    .groupby(["linha", "ano"], as_index=False)
    .agg(
        soma_passageiros=("total_passageiros", "sum"),
        total_registros=("linha", "size")
    )
)

# Pivot da soma de passageiros
piv_pass = (
    resumo_anos.pivot(index="linha", columns="ano", values="soma_passageiros")
       .rename(columns={2016: "passageiros_2016", 2018: "passageiros_2018"})
)

# Pivot da contagem de registros (nomes pedidos)
piv_regs = (
    resumo_anos.pivot(index="linha", columns="ano", values="total_registros")
       .rename(columns={2016: "total_2016", 2018: "total_2018"})
)


df_comparacoes_anos = (
    piv_pass.join(piv_regs, how="outer")
            .reset_index()
            .fillna(0)
)

for c in ["passageiros_2016", "passageiros_2018", "total_2016", "total_2018"]:
    df_comparacoes_anos[c] = df_comparacoes_anos[c].round().astype("Int64")


df_comparacoes_anos["perc_diferenca_passageiros"] = np.where(
    df_comparacoes_anos["passageiros_2016"] > 0,
    df_comparacoes_anos["passageiros_2018"] / df_comparacoes_anos["passageiros_2016"],
    np.nan
)

df_comparacoes_anos["perc_diferenca_registros"] = np.where(
    df_comparacoes_anos["total_2016"] > 0,
    df_comparacoes_anos["total_2018"] / df_comparacoes_anos["total_2016"],
    np.nan
)

df_comparacoes_anos = df_comparacoes_anos.sort_values(by='perc_diferenca_registros', ascending=False)

df_comparacoes_anos.head()
```

## Validação de linhas com registros contínuos
Dentro do novo período de análise, vamos verificar as linhas de ônibus que possuem registros todos os dias.

```python
df_chk = df_correcoes[["linha", "data"]].copy()
df_chk = df_chk.dropna(subset=["linha", "data"]).drop_duplicates(["linha", "data"])

# Janela global da análise
inicio_analise =  df_chk["data"].min()
fim_analise = df_chk["data"].max()
datas_esperadas = pd.date_range(inicio_analise, fim_analise, freq="D")
total_dias_periodo = len(datas_esperadas)

# Contagem de registros por ano (dias únicos por linha)
df_chk["ano"] = df_chk["data"].dt.year
reg_por_ano = (
    df_chk.groupby(["linha", "ano"])["data"]
    .nunique()
    .unstack(fill_value=0)
)

# Anos completos dentro da janela (evita punir ano parcial no início/fim)
anos_completos = [
    ano for ano in range(inicio_analise.year, fim_analise.year + 1)
    if pd.Timestamp(year=ano, month=1, day=1) >= inicio_analise
    and pd.Timestamp(year=ano, month=12, day=31) <= fim_analise
]

# Critério 1: pelo menos 365 registros em cada ano completo
if anos_completos:
    crit_365 = (reg_por_ano.reindex(columns=anos_completos, fill_value=0) >= 365).all(axis=1)
else:
    crit_365 = pd.Series(True, index=reg_por_ano.index)

# Critério 2: cobertura literal completa entre início e fim (sem faltar nenhum dia)
crit_cobertura_total = (
    df_chk.groupby("linha")["data"].nunique().eq(total_dias_periodo)
)

# Série final com as linhas que passam nos dois critérios
linhas_completas = pd.Series(
    sorted(crit_365.index[crit_365 & crit_cobertura_total]),
    name="linha"
)

linhas_completas
```

## Modelos estudados

Neste estudo serão utilizadas duas bibliotecas de aprendizado de máquina, cada uma utilizando diferentes técnicas de previsão de valores futuros, após o treinamento serão feitas comparações destas para validação do modelo que mais se encaixa no problema proposto.   

- `NeuralProphet`: Esta biblioteca utiliza técnicas de deep learning para a busca de padrões em séries temporais. Desta forma conseguindo entender padrões de tendências e sazonalidades.  Esta é uma variação da clássica biblioteca Prophet, porém por baixo dos panos utiliza o PyTorch para os processos de deep learning, esta foi criada por Oskar Triebe em colaboração com outros perquisidores na universidade de Stanford. Esta biblioteca por padrão tem a capacidade de realizar previsões de multi-séries (Modelo global), este recurso sendo de grande importância para o problema proposto, já que será necessário fazer uma previsão para cada linha de ônibus, sem a necessidade de criar um modelo para cada uma.
Nesta biblioteca já existe um recurso de cálculo de lag (Atraso) de registros de forma nativa, não sendo necessário a criação de campos de lag manuais.

- `Lightgbm`: A Light Gradient Boosting Machine (Lightgbm) é uma biblioteca desenvolvida pelos pesquisadores Guolin Ke e Tie-Yan Liu da Microsoft Research, está possui a intenção de ser uma evolução da GXBoost com otimização de consumo de RAM e outras melhorias. Para o caso estudado será utilizado a classe LGBMRegressor, especializada em problemas de regressão. Esta utiliza a técnica de Gradient Boosting. Esta classe também possui a capacidade do trabalho com múlti-séries utilizando um modelo global. Esta biblioteca exige mais engenharia de features, como por exemplo a adição de dados de lag para melhor previsão de sazonalidades.



## Engenharia de features

Abaixo é realizado  a engenharia de features, adicionando informações uteis para o treinamento.

### Dados de calendário:
- mês, dia do mês, ano.
- férias escolares
- pandemia de covid e lockdowns (essencial para o modelo entender a queda de uso durante o início da pandemia.

### Lags e médias moveis
As defasagens de dados e médias móveis são features de grande ajuda para melhor performance do modelo Lightgbm, sendo os lags de 1, 7, 14 e 28 dias e médias móveis de 7 e 30 dias.  


```python
#Engenharia de features para o treinamento/testes

print(len(df_final))

df_final = df_final.sort_values(by=['linha', 'data']).reset_index(drop=True)

# Dado de calendário
df_final['mes'] = df_final['data'].dt.month
df_final['dia_do_mes'] = df_final['data'].dt.day
df_final['ano'] = df_final['data'].dt.year

# Férias escolares no Brasil (Janeiro e Julho)
df_final['ferias_escolares'] = df_final['mes'].isin([1, 7]).astype(int)

# Período de covid definido em São Paulo e maiores lockdowns
df_final['pandemia_covid'] = ((df_final['data'] >= '2020-03-16') & (df_final['data'] <= '2023-05-12'))
filtro_lockdown = (
    ((df_final['data'] >= '2020-03-24') & (df_final['data'] <= '2020-05-31')) |
    ((df_final['data'] >= '2021-03-15') & (df_final['data'] <= '2021-04-11'))
)
df_final['lockdown_covid'] = filtro_lockdown


group_linha = df_final.groupby('linha')['total_passageiros']



# Lags diários e semanais
df_final['lag_1d'] = group_linha.shift(1)   # Ontem
df_final['lag_7d'] = group_linha.shift(7)   # Mesmo dia da semana passada
df_final['lag_14d'] = group_linha.shift(14) # Mesmo dia há 2 semanas
df_final['lag_28d'] = group_linha.shift(28) # Mesmo dia há 4 semanas (aprox. 1 mês)

# médias móveis
df_final['media_movel_7d'] = (
    df_final.groupby('linha')['total_passageiros']
    .transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).mean())
)

df_final['media_movel_30d'] = (
    df_final.groupby('linha')['total_passageiros']
    .transform(lambda x: x.shift(1).rolling(window=30, min_periods=1).mean())
)

df_final['desvio_padrao_7d'] = (
    df_final.groupby('linha')['total_passageiros']
    .transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).std())
)
```

```python
# Consolidação para NeuralProphet: garante 1 registro por linha e data
df_treinamento_np = df_treinamento.copy()
df_treinamento_np['data'] = pd.to_datetime(df_treinamento_np['data'])

dups_id_data = df_treinamento_np.duplicated(subset=['linha', 'data']).sum()
print(f'Duplicatas linha+data antes da consolidação: {dups_id_data}')

if dups_id_data > 0:
    agg_map = {
        'tipo': 'first',
        'area': 'first',
        'empresa': 'first',
        'feriado': 'max',
        'total_passageiros': 'sum',
        'dia_semana': 'first',
        'mes': 'first',
        'dia_do_mes': 'first',
        'ano': 'first',
        'ferias_escolares': 'first',
        'pandemia_covid': 'max',   # Mantém o valor booleano máximo (True se houver ocorrência)
        'lockdown_covid': 'max',   # Mantém o valor booleano máximo
    }

    df_treinamento_np = (
        df_treinamento_np
        .groupby(['linha', 'data'], as_index=False)
        .agg(agg_map)
    )

# Recalcula lags e estatísticas móveis sobre a série consolidada
df_treinamento_np = df_treinamento_np.sort_values(['linha', 'data']).reset_index(drop=True)
grp = df_treinamento_np.groupby('linha')['total_passageiros']

df_treinamento_np['lag_1d'] = grp.shift(1)
df_treinamento_np['lag_7d'] = grp.shift(7)
df_treinamento_np['lag_14d'] = grp.shift(14)
df_treinamento_np['lag_28d'] = grp.shift(28)

df_treinamento_np['media_movel_7d'] = (
    grp.transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).mean())
)
df_treinamento_np['media_movel_30d'] = (
    grp.transform(lambda x: x.shift(1).rolling(window=30, min_periods=1).mean())
)
df_treinamento_np['desvio_padrao_7d'] = (
    grp.transform(lambda x: x.shift(1).rolling(window=7, min_periods=1).std())
)

# Remove linhas iniciais sem histórico mínimo de lags
df_treinamento_np = df_treinamento_np.dropna().copy()

# NeuralProphet não aceita dtype pandas nullable (Int64) ou booleanos puros de forma nativa em algumas versões,
# então convertemos todos os regressores e a variável alvo para float64 padrão
colunas_numericas_np = [
    'feriado', 'total_passageiros', 'dia_semana', 'mes', 'dia_do_mes', 'ano',
    'ferias_escolares', 'pandemia_covid', 'lockdown_covid'
]
df_treinamento_np[colunas_numericas_np] = (
    df_treinamento_np[colunas_numericas_np]
    .apply(pd.to_numeric, errors='coerce')
    .astype('float64')
)

print(f'Tamanho após consolidação e recálculo: {len(df_treinamento_np)}')
print('Duplicatas linha+data depois:', df_treinamento_np.duplicated(subset=['linha', 'data']).sum())

# Sobrescreve a base usada na célula de modelagem
df_treinamento = df_treinamento_np.copy()
```

```python
# Compatibilidade temporária para NeuralProphet + pandas em ambientes web como google colab
import pandas as pd
import numpy as np

# 1) NeuralProphet 0.9.0 usa Series.view() em alguns fluxos
if not hasattr(pd.Series, 'view'):
    def _series_view(self, dtype=None):
        if dtype is None:
            return self.to_numpy().view()
        return self.astype(dtype, copy=False)
    pd.Series.view = _series_view
    print('Patch aplicado: pd.Series.view')

# 2) Reset explícito de DataFrameGroupBy.apply para base do pandas
try:
    from pandas.core.groupby.generic import DataFrameGroupBy
    from pandas.core.groupby.groupby import GroupBy

    DataFrameGroupBy.apply = GroupBy.apply
    if hasattr(DataFrameGroupBy, '_np_orig_apply'):
        delattr(DataFrameGroupBy, '_np_orig_apply')
except Exception as exc:
    print(f'Não foi possível resetar DataFrameGroupBy.apply: {exc}')

# 3) Patch seguro para preservar coluna ID quando ela está no índice do resultado
try:
    from pandas.core.groupby.generic import DataFrameGroupBy

    DataFrameGroupBy._np_orig_apply = DataFrameGroupBy.apply

    def _apply_compat(self, func, *args, **kwargs):
        res = DataFrameGroupBy._np_orig_apply(self, func, *args, **kwargs)
        try:
            if isinstance(res, pd.DataFrame) and ('ID' not in res.columns):
                if isinstance(res.index, pd.MultiIndex) and ('ID' in res.index.names):
                    res = res.copy()
                    res['ID'] = res.index.get_level_values('ID')
        except Exception:
            pass
        return res

    DataFrameGroupBy.apply = _apply_compat
    print('Patch aplicado: compatibilidade de groupby.apply para coluna ID')
except Exception as exc:
    print(f'Não foi possível aplicar patch de groupby.apply: {exc}')
```

```python
import warnings
warnings.filterwarnings("ignore")

import os
import pickle
from pathlib import Path
import pandas as pd
import numpy as np
import torch
import logging
from sklearn.metrics import mean_absolute_error, mean_squared_error
from neuralprophet import NeuralProphet

# Desativar logs excessivos
logging.getLogger("NP").setLevel(logging.ERROR)
logging.getLogger("pytorch_lightning").setLevel(logging.ERROR)
logging.getLogger("lightning").setLevel(logging.ERROR)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Dispositivo selecionado para o NeuralProphet Global: {DEVICE}")

# 1) Preparação da Base Unificada no padrão NeuralProphet Global
df_modelo = df_treinamento.copy()
df_modelo['data'] = pd.to_datetime(df_modelo['data'])

# Para modelos globais no NeuralProphet, o identificador da série precisa estar na coluna 'ID'
df_global = df_modelo.rename(columns={
    'data': 'ds',
    'total_passageiros': 'y',
    'linha': 'ID'
}).copy()

regressoras = ['feriado', 'ferias_escolares', 'pandemia_covid', 'lockdown_covid']
colunas_necessarias = ['ds', 'y', 'ID'] + regressoras
df_global = df_global[colunas_necessarias].sort_values(['ID', 'ds']).reset_index(drop=True)

# Filtrar séries com histórico mínimo aceitável
tamanhos = df_global.groupby('ID').size()
ids_elegiveis = tamanhos[tamanhos >= 120].index
df_global = df_global[df_global['ID'].isin(ids_elegiveis)].reset_index(drop=True)

print(f"Total de linhas no dataset global: {len(df_global)}")
print(f"Quantidade de séries (IDs) únicas: {df_global['ID'].nunique()}")

# 2) Divisão de Treino e Teste temporal global
train_end = pd.Timestamp('2025-12-31')
test_end = pd.Timestamp('2026-06-30')

df_treino = df_global[df_global['ds'] <= train_end].copy()
df_teste = df_global[(df_global['ds'] > train_end) & (df_global['ds'] <= test_end)].copy()

# 3) Criação do NeuralProphet Global
modelo_np_global = NeuralProphet(
    yearly_seasonality=True,
    weekly_seasonality=True,
    daily_seasonality=False,
    n_lags=28,
    n_forecasts=1,
    epochs=50,
    learning_rate=0.001,
    batch_size=512, # Lotes maiores para otimizar GPU
    loss_func="Huber",
    accelerator="gpu" if torch.cuda.is_available() else "cpu",
    trainer_config={
        "logger": False,
        "enable_progress_bar": True,
        "enable_model_summary": False,
    }

)

for r in regressoras:
    modelo_np_global = modelo_np_global.add_future_regressor(r)

# Ajustar modelo de forma global (processamento massivo na GPU)
print("Iniciando treinamento do NeuralProphet Global...")
metrics = modelo_np_global.fit(df_treino, freq='D', progress='bar', early_stopping=True)
print("Treinamento concluído!")

# 4) Predição Global no Intervalo de Teste
print("Realizando previsões globais...")
df_completo_para_predict = pd.concat([df_treino, df_teste], ignore_index=True)
fcst_global = modelo_np_global.predict(df_completo_para_predict)

fcst_teste = fcst_global[fcst_global['ds'].isin(df_teste['ds'])].copy()
pred_col = 'yhat1' if 'yhat1' in fcst_teste.columns else 'yhat'

# Avaliação de métricas globais
prevs = df_teste[['ds', 'ID', 'y']].merge(
    fcst_teste[['ds', 'ID', pred_col]],
    on=['ds', 'ID'],
    how='left'
).rename(columns={pred_col: 'y_pred', 'ID': 'linha'})

prevs['y_pred'] = prevs['y_pred'].clip(lower=0).fillna(0)

# Estatísticas globais do NeuralProphet
y_true = prevs['y'].to_numpy()
y_pred = prevs['y_pred'].to_numpy()

mae_glob = mean_absolute_error(y_true, y_pred)
rmse_glob = np.sqrt(mean_squared_error(y_true, y_pred))
mape_glob = np.mean(np.abs((y_true - y_pred) / np.maximum(y_true, 1))) * 100

df_resultados_metricas = pd.DataFrame([{
    'cenario': 'NeuralProphet_Global',
    'MAE': mae_glob,
    'RMSE': rmse_glob,
    'MAPE_%': mape_glob,
    'linhas_treinadas': prevs['linha'].nunique(),
    'n_amostras_teste_total': len(prevs)
}])

# Métricas individuais por linha para compatibilidade com relatórios posteriores
metricas_linhas = []
for linha_id, grp in prevs.groupby('linha'):
    yt = grp['y'].to_numpy()
    yp = grp['y_pred'].to_numpy()
    m_mae = mean_absolute_error(yt, yp)
    m_rmse = np.sqrt(mean_squared_error(yt, yp))
    m_mape = np.mean(np.abs((yt - yp) / np.maximum(yt, 1))) * 100
    metricas_linhas.append({
        'linha': linha_id,
        'MAE': m_mae,
        'RMSE': m_rmse,
        'MAPE_%': m_mape
    })

metricas_por_linha_cenario = {'NeuralProphet_Global': pd.DataFrame(metricas_linhas)}
resultados_previsoes = {'NeuralProphet_Global': prevs}
cenario_unico = {'nome': 'NeuralProphet_Global'}

# Salvar modelo global único
modelo_dir = Path.cwd() / 'modelo'
modelo_dir.mkdir(parents=True, exist_ok=True)
with open(modelo_dir / 'neuralprophet_global.pkl', 'wb') as f:
    pickle.dump(modelo_np_global, f)

print("NeuralProphet Global finalizado e salvo!")
display(df_resultados_metricas)
```

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import lightgbm as lgb
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error

print("=== Preparando dados para o LightGBM Global ===")

# 1) Preparação dos dados utilizando as features já construídas
# Usaremos a base df_final que contém os lags e médias móveis já calculados
df_lgb = df_final.copy()
df_lgb['data'] = pd.to_datetime(df_lgb['data'])

# Definir as features (recursos) para o modelo
features_calendario = ['dia_semana', 'mes', 'dia_do_mes', 'ano', 'feriado', 'ferias_escolares', 'pandemia_covid', 'lockdown_covid']
features_lags = ['lag_1d', 'lag_7d', 'lag_14d', 'lag_28d']
features_estatisticas = ['media_movel_7d', 'media_movel_30d', 'desvio_padrao_7d']

# Tratamento de variáveis categóricas (ID da linha / 'linha')
df_lgb['linha_cat'] = df_lgb['linha'].astype('category')
features_categoricas = ['linha_cat']

features_modelo = features_calendario + features_lags + features_estatisticas + features_categoricas
target = 'total_passageiros'

# Garantir que não há NaNs nas features que o LightGBM precisa treinar
# O LightGBM lida nativamente com NaNs, mas remover os NaNs iniciais dos lags evita distorções de escala inicial
df_lgb_limpo = df_lgb.dropna(subset=[target] + features_lags).reset_index(drop=True)

# 2) Divisão de Treino e Teste Temporal coerente com o NeuralProphet
train_end = pd.Timestamp('2025-12-31')
test_end = pd.Timestamp('2026-06-30')

df_treino_lgb = df_lgb_limpo[df_lgb_limpo['data'] <= train_end].copy()
df_teste_lgb = df_lgb_limpo[(df_lgb_limpo['data'] > train_end) & (df_lgb_limpo['data'] <= test_end)].copy()

X_train = df_treino_lgb[features_modelo]
y_train = df_treino_lgb[target].astype(float)
X_test = df_teste_lgb[features_modelo]
y_test = df_teste_lgb[target].astype(float)

print(f"Registros de Treino: {len(X_train)} | Teste: {len(X_test)}")

# 3) Treinamento do Modelo LightGBM
model_lgb = lgb.LGBMRegressor(
    n_estimators=150,
    learning_rate=0.08,
    num_leaves=63,
    random_state=42,
    n_jobs=-1
)

print("Treinando LightGBM Global...")
model_lgb.fit(
    X_train,
    y_train,
    categorical_feature=features_categoricas
)
print("Treinamento concluído!")

# 4) Predições e Avaliação
preds = model_lgb.predict(X_test)
df_teste_lgb['y_pred'] = np.clip(preds, a_min=0, a_max=None)

# Métricas Globais
y_true_lgb = df_teste_lgb[target].to_numpy()
y_pred_lgb = df_teste_lgb['y_pred'].to_numpy()

mae_lgb = mean_absolute_error(y_true_lgb, y_pred_lgb)
rmse_lgb = np.sqrt(mean_squared_error(y_true_lgb, y_pred_lgb))
mape_lgb = np.mean(np.abs((y_true_lgb - y_pred_lgb) / np.maximum(y_true_lgb, 1))) * 100

df_resultados_lgb = pd.DataFrame([{
    'cenario': 'LightGBM_Global',
    'MAE': mae_lgb,
    'RMSE': rmse_lgb,
    'MAPE_%': mape_lgb,
    'linhas_treinadas': df_teste_lgb['linha'].nunique(),
    'n_amostras_teste_total': len(df_teste_lgb)
}])

print("\n=== Resultado de Métricas Globais ===")
display(df_resultados_lgb)

```

```python
# ===== Relatório 3: gráficos por linha para o LightGBM (Último 1 Ano / Período de Teste) =====
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error

if "df_teste_lgb" not in globals() or df_teste_lgb.empty:
    raise ValueError("O DataFrame de teste do LightGBM 'df_teste_lgb' não foi encontrado ou está vazio.")

prevs_lgb = df_teste_lgb[['data', 'linha', 'total_passageiros', 'y_pred']].copy()
prevs_lgb['data'] = pd.to_datetime(prevs_lgb['data'])

# 1) Calcular métricas por linha especificamente para o LightGBM
metricas_linhas_lgb = []
for linha_id, grp in prevs_lgb.groupby('linha'):
    yt = grp['total_passageiros'].to_numpy()
    yp = grp['y_pred'].to_numpy()

    m_mae = mean_absolute_error(yt, yp)
    m_rmse = np.sqrt(mean_squared_error(yt, yp))
    m_mape = np.mean(np.abs((yt - yp) / np.maximum(yt, 1))) * 100

    metricas_linhas_lgb.append({
        'linha': linha_id,
        'MAE': m_mae,
        'RMSE': m_rmse,
        'MAPE_%': m_mape
    })

df_met_linha_lgb = pd.DataFrame(metricas_linhas_lgb)

# 2) Ordenar pelo MAPE para selecionar os 5 melhores e 5 piores
df_ordenado_lgb = df_met_linha_lgb.sort_values("MAPE_%")
melhores_linhas_lgb = df_ordenado_lgb["linha"].drop_duplicates().head(5).tolist()
piores_linhas_lgb = df_ordenado_lgb["linha"].drop_duplicates().tail(5).tolist()

linhas_plot_lgb = melhores_linhas_lgb + piores_linhas_lgb

# Mapear métricas para títulos
dict_metricas_lgb = {}
for _, row in df_met_linha_lgb.iterrows():
    dict_metricas_lgb[row['linha']] = {
        'RMSE': row['RMSE'],
        'MAE': row['MAE'],
        'MAPE': row['MAPE_%']
    }

# 3) Plotar comparativo para as 10 linhas selecionadas
fig, axes = plt.subplots(len(linhas_plot_lgb), 1, figsize=(16, 4 * len(linhas_plot_lgb)), sharex=False)

if len(linhas_plot_lgb) == 1:
    axes = [axes]

for i, (ax, linha_id) in enumerate(zip(axes, linhas_plot_lgb)):
    df_linha = prevs_lgb[prevs_lgb["linha"] == linha_id].copy().sort_values("data")

    # Filtrar o último 1 ano da série disponível
    data_fim = df_linha["data"].max()
    data_ini = data_fim - pd.Timedelta(days=365)
    df_1ano = df_linha[df_linha["data"] >= data_ini].copy()

    if df_1ano.empty:
        df_1ano = df_linha

    ax.plot(df_1ano["data"], df_1ano["total_passageiros"], label="y (real)", linewidth=2, color='tab:blue')
    ax.plot(df_1ano["data"], df_1ano["y_pred"], label="y_pred (LightGBM)", linewidth=2, alpha=0.9, color='tab:orange')

    met_str = ""
    if linha_id in dict_metricas_lgb:
        m = dict_metricas_lgb[linha_id]
        met_str = f" (RMSE: {m['RMSE']:.2f}, MAE: {m['MAE']:.2f}, MAPE: {m['MAPE']:.2f}%)"

    grupo_label = "Melhor" if i < 5 else "Pior"
    ax.set_title(f"[{grupo_label} #{i%5 + 1}] Linha {linha_id}{met_str} | Janela: {df_1ano['data'].min().date()} até {df_1ano['data'].max().date()}")
    ax.set_ylabel("Passageiros")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="best")

axes[-1].set_xlabel("Data")
plt.suptitle("Comparativo y vs y_pred - 5 Melhores vs 5 Piores por MAPE (LightGBM Global)", y=1.005, fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
